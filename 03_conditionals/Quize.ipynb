```python
# 3-1 조건문의 기본 문법

temperature = 38.5
if temperature >=37.5:               # if문 블록
    print("발열이 있습니다.")
    print("해열제를 처방합니다.")
print("체온 확인 완료")                # if문 블록 밖: 들여 쓰기 없음 ->  항상 실행

print()

age = 16
if age >= 18:
    print("성인용 약물을 처방합니다.")
else:
    print("소아용 약물을 처방합니다.")
```

```python
# if/elif/else는 위에서부터 차례로 확인하고, 처음으로 True인 블록 하나만 실행

bmi = 27.5

if bmi < 18.5:
    status = "저체중"
elif bmi < 25.0:
    status = "정상"
elif bmi < 30.0:
    status = "과체중"
else:
    status = "비만"
print(f"BMI: {bmi} → {status}")
```

```python
# 조건문 안에 조건문을 넣는 중첩 if도 가능하지만, 읽기 어려워지므로 보통 2단계만 사용

spo2 = 92
has_mask = True

if spo2 < 95:
    if has_mask:
        print("고농도 산소 투여")
    else:
        print("마스크 먼저 착용 후 산소 투여")
else:
    print("산소 처치 불필요")
    
```
