# 연습문제 3. 환자 응급도 분류하기

- 아래 환자의 생체징후로 응급도를 분류하고 처치 방향과 대기시간을 출력하세요.

| 분류 | 기준 | 처치 방향 | 대기 시간 |
|---|---|---|---|
| 즉시 처치 | 의식 없고, 수축기혈압 >= 180 또는 체온 >= 39.0 | 즉시 응급처치 | 0분 |
| 응급 처치 | 통증 8점 이상 또는 호흡곤란 | 응급 처치 후 정밀검사 | 15분 |
| 준응급 | 통증 4-7점 또는 발열(37.5-38.9) | 약물 처치 후 전문의 진료 | 30분 |
| 비응급 | 위에 해당하지 않은 | 일반 진료 | 60분 |

## 출력결과 예시

=== 환자 응급도 분류 결과 ===
환자명: 김환자 (45세)
생체징후: 혈압 160/90, 체온 38.2도, 통증 3점
의식상태: 의식 있음, 호흡곤란: 호흡곤란 없음
🟡 분류 결과: 준응급
📋 처치 방향: 약물 처치 후 전문의 진료
⏰ 예상 대기시간: 30분

```python
# 주어진 환자 데이터

patient = {
    "name": "김환자", 
    "age": 45, 
    "systolic_bp": 160, 
    "diastolic_bp":90, 
    "temperature": 38.2,
    "pain_level": 3, 
    "is_conscious": True, 
    "has_breathing_difficulty": False,
}
```

```python
# 만들어야 하는 변수 : 응급 분류, 처치 방향,  대기 시간

# 주어진 조건을 변수로 만들기
name = patient["name"]
age = patient["age"]
systolic_bp = patient["systolic_bp"]
diastolic_bp = patient["diastolic_bp"]
temperature = patient["temperature"]
pain_level = patient["pain_level"]
is_conscious = patient["is_conscious"]
dyspnea = patient["has_breathing_difficulty"]

# 분류 변수 만들기
if not is_conscious or systolic_bp >= 180 or temperature >= 39.0:
    emergency = "즉시 처치"
    treatment = "즉시 응급처치"
    wait_time = "0분"
elif pain_level >= 8 or dyspnea:
    emergency = "응급 처치"
    treatment = "응급 처치 후 정밀검사"
    wait_time = "15분"
elif (4 <= pain_level <= 8) or (37.5 <= temperature < 39.0):
    emergency = "준응급"
    treatment = "약물 처치 후 전문의 진료"
    wait_time = "30분"
else:
    emergency = "비응급"
    treatment = "일반 진료" 
    wait_time = "60분"

# 의식 상태 조건문
if is_conscious == True:
    conscious = "의식 있음"
else:
    conscious = "의식 없음"

# 호흡곤란 상태 조건문
if dyspnea == True:
    dyspnea = "호흡곤란 있음"
else:
    dyspnea = "호흡곤란 없음"

# 출력하기
print("=== 환자 응급도 분류 결과 ===")
print(f"환자명: {name} ({age}세)")
print(f"생체징후: 혈압 {systolic_bp}/{diastolic_bp}, 체온 {temperature}도, 통증 {pain_level}점")
print(f"의식상태: {conscious}, 호흡곤란: {dyspnea}")
print(f"🟡 분류 결과: {emergency}")
print(f"📋 처치 방향: {treatment}")
print(f"⏰ 예상 대기시간: {wait_time}")
```
