```python
# 3-1 조건문의 기본 문법

temperature = 38.5
if temperature >=37.5:               # if문 블록
    print("발열이 있습니다.")
    print("해열제를 처방합니다.")
print("체온 확인 완료")                # if문 블록 밖: 들여 쓰기 없음 ->  항상 실행

print()

age = 16
if age >= 18:
    print("성인용 약물을 처방합니다.")
else:
    print("소아용 약물을 처방합니다.")
```

```python
# if/elif/else는 위에서부터 차례로 확인하고, 처음으로 True인 블록 하나만 실행

bmi = 27.5

if bmi < 18.5:
    status = "저체중"
elif bmi < 25.0:
    status = "정상"
elif bmi < 30.0:
    status = "과체중"
else:
    status = "비만"
print(f"BMI: {bmi} → {status}")
```

```python
# 조건문 안에 조건문을 넣는 중첩 if도 가능하지만, 읽기 어려워지므로 보통 2단계만 사용

spo2 = 92
has_mask = True

if spo2 < 95:
    if has_mask:
        print("고농도 산소 투여")
    else:
        print("마스크 먼저 착용 후 산소 투여")
else:
    print("산소 처치 불필요")
    
```

```python
# 조건 확인

BP_NORMAL = (90, 139) # 정상혈압
bp = 135
if BP_NORMAL[0] <= bp <= BP_NORMAL[1]:
    print("혈압 정상 범위")

symptoms = ["발열", "기침", "두통"]
if "호흡곤란" not in symptoms:
    print("호흡곤란 없음")

if "고혈압" in "고혈압성 심장병":
    print("혈압약 처방 검토")
```

```python
# Falsy 값: None, False, 0, " ", [], {}, set()은 조건에서 거짓으로 취급/ 확인하기

medications = []
if not medications:
    print("처방약 없음 → 처방 필요")

allergies = {"페니실린", "아스피린"}
new_rx = {"아스피린", "인슐린"}
conflict = allergies & new_rx
if conflict:
    print(f"알레르기 충돌! → {conflict}")
```

```python
# 딕셔너리와 결합

patient = {
    "name": "김환자",
    "age": 68,
    "diagnosis": "고혈압, 당뇨",
    "vitals":{
        "bp": 158,
        "temp": 37.8
    }
}

if patient["age"] >= 65:
    print("고령환자 → 낙상 위험 주의")
if patient["vitals"]["bp"] >= 140:
    print("고혈압 처치 필요")
if "당뇨" in patient["diagnosis"]:
    print("혈당 모니터링 필요")
```
