# 연산자

```python

a = 10
b = 3
print(a+b)  # 더하기
print(a-b)  # 빼기
print(a*b)  # 곱하기
print(a/b) # 나누기
print(a//b) # 몫
print(a%b)  # 나머지
print(a**b) # 제곱
```

```python
# 동전 교환( 500원, 100원, 50원, 10원, 1원)
money = 7777

# 500원 동전 개수
c500 = money // 500  # 500원 동전으로 나눈 몫
money = money % 500 # 500원 동전으로 나눈 나머지 금액
print("500원짜리: ", c500)

# 100원 동전 개수
c100 = money // 100 # 100원 동전으로 나눈 몫
money = money % 100 # 100원 동전으로 나눈 나머지 금액
print("100원짜리: ", c100)

# 50원 동전 개수
c50 = money // 50 # 50원 동전으로 나눈 몫
money = money % 50 # 50원 동전으로 나눈 나머지 금액
print("50원짜리: ", c50)

# 10원 동전 개수
c10 = money // 10 # 10원 동전으로 나눈 몫
money = money % 10 # 10원 동전으로 나눈 나머지 금액
print("10원짜리: ", c10)

# 1원 동전 개수
print("1원짜리: ", money) # 1원 동전 개수

# 모든 동전/돈의 합계
print("모든 동전/돈의 합계: ", c500*500 + c100*100 + c50*50 + c10*10 + money)
```

# 대입 연산자(assignment operator)

```python
student = 30
english_average = 80.2
student += 3
print(student)
```

```python
jenny_weight = 60
tom_weight = 80
print(jenny_weight >= tom_weight) # jenny_weight가 tom_weight보다 크거나 같은지 비교
print(jenny_weight <= tom_weight) # jenny_weight가 tom_weight보다 작거나 같은지 비교
```

```python
# 수학의 연산자 우선순위와 동일하게 연산
a = 10
b = 20
print(a + b * 3)
```

```python
# 문자열
print('첫번째줄\n두번째줄\n세번째줄')    # \n: 줄바꿈
print('첫번째줄\t두번째줄\t세번째탭키')   # \t: 탭키
print('첫번째줄', '두번째줄\b뒤로한칸이동')  # \b: 뒤로 한 칸 이동
```

```python
# 데이터 타입 체크
a = 10
b = 3.14
c = 'Hello World!'

print(type(a)) # <class 'int'>
print(type(b)) # <class 'float'>
print(type(c))  # <class 'str'>
```

```python
# 문자열 추출: 첫글짜는 인덱스 0부터 시작, 마지막 글자는 -1부터 시작
s = 'Hello World!'
print(s[0])  # H
print(s[4])  # o
print(s[-1]) #!
print(s[:])  # 처음부터 끝까지
print(s[0:4]) # 0~3까지 추출
```
