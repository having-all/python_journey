## 🐍 Quiz 1 — 클래스 활용하기

다음 조건을 만족하는 Student 클래스를 완성하세요.

```
1. __init__()을 이용해 학생의 이름(name)과 점수(score)를 저장하세요.
2. add_score() 메서드를 만들어 전달받은 점수만큼 기존 점수를 증가시키세요.
3 get_score() 메서드는 현재 점수를 반환하도록 작성하세요.
```


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def add_score(self, new_score):
        self.score += new_score
    
    def get_score(self):
        return self.score


student = Student("Kim", 80)
student.add_score(15)

print(student.get_score())
````

## 🐍 Quiz 2 — 딕셔너리 + 반복문 오류 수정하기
다음 코드는 학생별 점수를 저장한 딕셔너리에서 80점 이상인 학생의 이름만 출력하려는 코드입니다.
하지만 실행하면 오류가 발생합니다.
오류가 발생하는 부분을 찾아 정상적으로 실행되도록 코드를 수정하세요.

````python
scores = {
    "Kim": 85,
    "Lee": 72,
    "Park": 90,
    "Choi": 65
}

for name, score in scores.items():
    if score >= 80:
        print(name)
````

## 🐍 Quiz 2 — 클래스 + 리스트 + 딕셔너리 종합 실습 (선택)
다음은 여러 학생의 성적을 관리하는 StudentManager 클래스입니다.  
80점 이상인 학생만 찾아 이름과 점수를 딕셔너리로 반환하는 메서드를 완성하세요.

```
- __init__()에서 학생 정보를 리스트로 저장하세요.
- get_high_scores() 메서드를 작성하세요.
- for 반복문과 if 조건문을 사용하세요.
- 80점 이상인 학생만 {이름: 점수} 형태의 딕셔너리에 저장하세요.
- 최종 딕셔너리를 return으로 반환하세요.
- 리스트 컴프리헨션은 사용하지 마세요.
```



````python
class StudentManager:
    def __init__(self, students):
        self.students = students

    def get_high_scores(self):
        high_scores = {}                                           # 80점이상 학생을 담을 딕셔너리를 만든다
        for student in self.students:                              # 1명의 학생을 학생들의 목록에서 가져온다.
            if student['score'] >= 80:                             # 어떤 학생이냐면 점수가 80점 이상인 학생
                high_scores[student['name']] = student['score']    # 80점 이상 학생을 {이름:점수}로 딕셔너리에 담기
        return high_scores                                         # 80점 이상 학생 결과 반환

students = [
    {"name": "Kim", "score": 85},
    {"name": "Lee", "score": 72},
    {"name": "Park", "score": 90},
    {"name": "Choi", "score": 65},
    {"name": "Jung", "score": 80}
]

manager = StudentManager(students)
print(manager.get_high_scores())
````
