# 컬렉션

## 1. 컬렉션(Collection)
    1) '모음'이라는 뜻
    2) 데이터들을 모아두는 집합(저장소)
    3) 여러 값을 하나의 이름으로 묶어서 관리하는 자료형
    4) 파이썬에서 주로 사용되는 4가지 기본 컬렉션
        - 리스트, 튜플, 세트, 딕셔너리
    5) 컬렉션 자료형을 활용하면 데이터를 가공하고 저장하기 수월함

## 2. 리스트(List)
    1) 순서가 있는 변경 가능한 시퀀스 자료형(추가, 수정 삭제)
    2) 중복 값 허용
    3) 대괄호로 생성 or list()함수 사용
    4) 예시
        list_example = ["사과", "배", "단감", "포도", 100, 12.3]
        - 저장하려고 하는 값들의 자료형이 달라도 한 리스트에 저장 가능
        - C, Java에 비하면 큰 장점

## 3. 튜플(Tuple)
    1) 순서가 있는 변경 불가능한 시퀀스 자료형
    2) 중복 값 허용
    3) 소괄호로 생성 or tuple() 함수 사용
    4) 예시
        tuple_example = ("사과", "배", "단감", "포도")

## 4. 세트(Set)
    1) 순서가 없고 변경 가능한 자료형
    2) 중복 값 저장할 수 없음
    3) 중괄호로 생성 or set() 함수 사용
    4) 예시
        set_example = {"사과", "배", "단감", "포도"}
        - 수학에서 집합과 동일한 개념

## 5. 딕셔너리(Dictionary)
    1) 순서가 없고 key와 value을 쌍으로 가진 데이터구조의 연속
    2) 중복된 key를 허용하지 않음
    3) 중괄호로 생성 or dict() 함수 사용
    4) 예시
        dict_example = {
            "국어" : 90,
            "수학": 100,
            "영어" : 80
        }

```python
# 리스트: 자료형이 달라도 한 리스트에 저장 가능
a = []
a = ['python', 'java', 1, 10, [9, 55]]
print(a)
```

```python
a = [1, 2, 3, 4, 5]
print(sum(a))  # sum은 리스안의 모든 값의 합계
print(a[1:3])  # 리스트 a의 2번째, 3번째 값을 출력
print(a[-1])   # 리스트 a의 마지막 값을 출력
print(a[2])    # 리스트 a의 3번째 값을 출력

a[0] = 10  # 리스트 a의 1번째 값을 10으로 변경
print(a)
```

```python
# 튜플

# 소괄호 생략 가능
tuple_example = 10, 20, 30
print(tuple_example)
print(type(tuple_example))

# 튜플은 읽기전용
t1 =(10, 20, 30, 40)
print(t1[0]+t1[3])  # 튜플 t1의 1번째 값과 4번째 값의 합

print(t1[1:3]) # 2번째, 3번째 값 출력

t1[0] = 50
print(t1[0])  # 오류: 튜플값은 변경 불가

```

```python
# 세트
# 세트는 중복값을 허용하지 않는다
a = [1, 2, 3, 5, 10, 10, 5]
b = set(a)
print(a)
print(b)
```

# 컬렉션 간 차이점

## 1. 인덱스 유무(Unorderd)
    1) 인덱스가 없으면 순서가 없음
        - 세트는 인덱스가 없음
    2) 문자열의 인덱싱, 슬라이싱과 동일하게 동작
## 2. 변경가능 여부(mutable)
    - 튜플만 변경 불가능
## 3. 중복 가능 여부
     1) 세트는 중복 불가능
     2) 딕셔너리의 key 값 중복 불가능

# 컬렉션 내장함수
1. 생성된 컬렉션 객체에 동작하는 함수
2. 리스트의 멤버 함수
| 함수명 | 형태 | 사용 |
|---|---|---|
| append | 값 | 리스트의 끝에 항목을 추가 |
| extend | 리스트 | 리스트를 연장 |
| insert | 인덱스, 값 | 리스트의 원하는 위치에 항목 추가 |
| remove | 값 | 리스트의 해당 값 삭제 |
| pop | 값 | 리스트의 해당 값 반환 후 삭제 |
| index | 값 | 찾고자 하는 값의 인덱스 반환 |
| sort() | | 리스트 자체를 내부적으로 정렬 |
| 값 in 리스트 | | 리스트 내에 해당 값이 있는지 확인 (boolean) |

3. 딕셔너리의 멤버 함수
| 함수명 | 사용 |
|---|---|
| keys() | 키만 반환 |
| values() | 값만 반환 |
| items() | 키, 값의 튜플을 반환 |
| update(딕셔너리) | 두 딕셔너리를 병합함(겹치는 키가 있다면, 변수로 전달되는 키 값이 덮어쓰기 됨) |
| del[딕셔너리'키'] | 키 삭제 |
| clear() in 딕셔너리 | 특정 키의 값이 있는지 확인



```python
# 리스트와 튜플
a = [1, 2, 3]
b = (1, 2, 3)
print(type(a))
print(type(b))

# a(리스트)의 1번째 값 변경
a[0] = 100
print(a)

# b(튜플)의 1번째 값 변경 => 오류 불가능함
b[0] = 100
print(b)
```

```python
# 내장함수

# append: 리스트 끝에 항목 추가
a = [1, 2, 3, 4, 5]
a.append(10)
print(a) # 추가된 10을 포함함

# extend: 리스트를 연장 
a = [1, 2, 3, 4, 5]
b = [6, 7, 8, 9, 10]
a.extend(b) 
print(a)    # 리스트 a = a + b

# insert: 리스트 원하는 위치에 항목 추가
a = [1, 2, 3, 4, 5]
a.insert(1, 40)  # 리스트 a의 2번째 위치에 40 추가
print(a)

# remove: 리스트의 해당값 삭제
a = [1, 2, 30, 30, 4, 5]
a.remove(2)  # 리스트 a중 값: 2를 가진 것 삭제
print(a)

# pop: 리스트의 해당 값 반환 후 삭제
a = [1, 2, 3, 4, 5]
d = a.pop(2)   # 인덱스 2에 있는 값 반환
print(a) # 리스트 a의 3번째 값이 삭제된 리스트 출력
print(d) # 삭제된 값 출력

# index: 찾는 값의 인덱스를 반환
a = [2, 6, 7, 9, 10]
print(a.index(9))  # 9는 4번째 값(인덱스: 3)

# 값 in 리스트: 리스트 내에 해당 값이 있는지 확인(boolean)으로 출력
a = [1, 2, 3, 4, 5, 10]
b = 10
c = b in a  # 리스트 a에 b의 값 10이 있는지 
print(c)

# sort(): 리스트 자체르 내부적으로 정렬
a = [9, 10, 7, 19, 1, 2, 20, 21, 7, 8]
a.sort(reverse=True) # True: 큰 값부터 정렬, False: 작은 값부터 정렬
print(a)
```

```python
# 딕셔너리 함수

student = {'학번':1000, '이름':'파이썬', '학과':'코딩학과'}

# keys(): 키만 반환
print(student.keys())

# values(): 값만 반환
print(student.values())

# items(): 키, 값을 튜플로 반환
print(student.items())

# update(딕셔너리): 두 딕셔너리를 병합함
student = {'학번':1000, '이름':'파이썬', '학과':'코딩학과'}
student.update({'학번':2000, '이름':'자바스크립트', '학과':'웹학과'})
print(student)

# del 딕셔너리[키]: key 삭제
del student['학번']
print(student)

# 값 in 딕셔너리: 특정 키의 값이 있는지 확인
print('파이썬' in student.values())
print('자바스크립트' in student.values())

# clear(): 딕셔너리의 모든 값을 초기화
student.clear()

```
