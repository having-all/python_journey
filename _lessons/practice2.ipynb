# 평균 계산 문자

```python
# 성적 리스트를 입력받아 평균을 계산하는 함수를 작성

def calc_average(score):
    result = 0
    for i in score:
        result += i
    return result / len(score)

score = [80, 90, 100, 70, 60]
calc_average(score)
```

# 각도를 분류하는 조건문 문제

```python
# 각에서 0도 초과 90도 미만은 예각, 90도는 직각, 90도 초과 180도 미만은 둔각 180도는 평각
# 각을 입력으로 받고 각도를 분류해서 출력하는 함수를 만드시오. 이때 각은 0~180까지만 입력됨

def determine_angle(num):
    if 0 < num < 90:
        print('예갹')
    elif num == 90 :
        print('직각')
    elif 90 < num < 180:
        print('둔각')
    elif num == 180:
        print('평각')
    else:
        print('잘못된 입력 값입니다.')


determine_angle(100)
```

# 책클래스 생성

```python
# 책 클래스를 만드시오
# 속성값은 제목, 저자, 페이지를 가짐

class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

harry_potter = Book("해리포터와 불의 잔", "J.K. 롤링", 309)
print(harry_potter.title)
print(harry_potter.author)
print(harry_potter.pages)
```

# 동물 클래스 생성

```python
# 동물 클래스를 만들고, 이를 상속받는 소 클래스를 만드시오
# 동물 클래스는 밥을 먹는 eat 메소드를 가지고, 소 클래스에는 울음소리를 내는 moo 메소드를 가진다.


class Animal:
    def eat(self):
        return '먹는다'

class Cow(Animal):
    def moo(self):
        return '울음소리를 낸다'

my_cow = Cow()
print(my_cow.eat())
print(my_cow.moo())   
```

# 은행 계좌 클래스 생성

```python
# 은행 계좌 클래스를 만드시오
# 은행에서 계좌를 개설하면 예금주, 잔액이 설정된다.
# 입금 메소드와 출금 메소드를 생성하시오. 단, 입금 금액은 양수여야 하며, 출금은 계좌 잔액 이상의 금액으로 출금할 수는 없다

class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, money):
        self.balance += money
        if self.balance < 0:
            print("입금 금액이 잘못되었습니다.")
        else:
            return self.balance

    def withdraw(self, money):
        self.balance -= money
        if self.balance < 0:
            print("잔액이 부족합니다.")
        else:
            return self.balance

kim = Account("Kim", 100000)

print(kim.name)
print(kim.balance)

kim.deposit(100000)
print(kim.balance)

kim.withdraw(100000)
print(kim.balance)
```
