# 소수 판별 문제

```python
# 자연수 값을 입력받아 소수인지 판별하는 함수르 짜기
# 소수: 1과 자기 자신 외에는 어떤 수로도 나눌 수 없는 자연수
# 2부터 입력받은 숫자 바로 전까지 나누어 0이 된다면 소수가 아님
# %: 나머지 확인하는 함수

def is_prime(num):
    if num == 1:
        return False
    else:
        for i in range(2, num):
            if num % i == 0:
                return False
        return True

is_prime(17)
```

# 애너그램 판별 문제

```python
# 애너그램이란 문자의 순서를 재배열 했을 때 같은 문자열
# 문자열 재배열하여 확인해야 하므로 2개의 변수가 필요

def is_anagram(str1, str2):
    if sorted(str1) == sorted(str2):
        return True
    else:
        return False

is_anagram("Hello World", "World Hello") # True
is_anagram("Hello Bob", "Hello Lisa")    # False
```

# 문자열 모음 갯수 세기

```python
# 입력한 문자열에 모음 갯수를 세는 함수 생성

def count_volume(str):
    count = 0
    vowel = "aeiouAEIOU" # 대문자, 소문자 모음 모두 작성
    for s in str:
        if s in vowel:
            count += 1
    return count

count_volume("Hellow World")
```

# 리스트 중복 제거

```python
# set()를 사용하지 말고 리스트에서 중복을 제거하는 함수를 작성하시오.
a = [1, 3, 5, 6, 7, 8, 8, 8, 10, 10]

def list_set(ex_list):
    removed = []
    for data in ex_list:
        if data not in removed:
            removed.append(data)
    return removed

list_set(a)
```

# 각 자리 수의 합 계산

```python
# 주어진 자연수의 각 자리의 수의 합을 계산하는 함수 생성

def sum_of_digits(num):
    total = 0
    for i in str(num):
        total += int(i)
    return total

sum_of_digits(3689)
```
