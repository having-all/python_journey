# 클래스, 객체

## 1. 클래스(class)
    1) 실세계의 것을 모델링해 속성(attribute)과 동작(method)를 갖는 데이터타입
        - 현실 세계의 다양한 사물을 클래스로 표현 가능, 일종의 '설계도'
        - 속성을 필드(filed), 멤버 변수(member variable)이라고도 함
    2) 파이썬에서 string, list, dict 등이 모두 클래스로 구현된 것
    3) 다루고자 하는 데이터(변수)와 연산(함수)를 하나로 캡슐화(encapsulation)


## 2. 객체
    1) 클래스로 생성되어 구체화된 것
        - 객체를 instance라고 함
    2) 파이썬에서 string, list, dict 등이 모두 클래스로 구현된 객체
    3) 클래스가 instance화 되어 메모리에 상주하는 상태
    4) 클래스가 붕어빵틀이라면, 객체는 실제로 찍어낸 붕어빵이라고 비유 가능
        - 한 클래스로부터 여러 개의 객체를 찍어낼 수 있음
    5) 파이썬은 객체지향 프로그램 언어(OOP; Object Oriented Programming)
        - 데이터를 객체로 취급하여 프로그램에 반영한 것
        - 객체와 객체의 상호작용으로 프로그램이 동작
        - 코드의 재사용성 높음, 직관적인 코드분석 가능, 개발 및 디버깅 속도 향상

## 3. 객체 생성하기
    1) 객체를 생성하기 위해서는 객체의 모체가 되는 클래스를 미리 선언해야 함
    2) 생성자 (init함수)
        - __init__(self)
        - 인스턴스를 생성하면 무조건 호출되는 메소드, 동시에 변수의 값을 초기화

            - 코드도 간결해지고 변수에 초기값을 대입하는 것을 잊어버리는 것을 방지
        - 객체가 생성될 때 자동으로 호출됨

        - self 외에 별도의 매개변수를 사용하지 않는 생성자를 기본 생성자라고 함
                - 매개변수가 있는 생성자
            - 생성자도 다른 메소드처럼 매개변수를 사용할 수 있다

## 4. 메소드(method)
    1) 멤버함수, 객체.함수명()으로 호출
    2) 해당 객체의 속성에 대한 연산 수행

```python
class Car:
    # 자동차의 속성(색상, 속도 등)
    color = " "
    speed = 0

    def __init__(self, color, speed):
        self.color = color
        self.speed = speed

    # 자동차의 동작(속도올리기, 속도 내리기)
    def upSpeed(self, value):
        self.speed += value
    def downSpeed(self, value):
        self.speed -= value 

myCar = Car('빨강', 10)
myCar.upSpeed(30)
myCar2 = Car('파랑', 20)
print("자동차 1의 색상은 %s이며, 현재속도는 %dkm/hr" % (myCar.color, myCar.speed))
print(f"자동차 1의 색상은 {myCar.color}이며, 현재속도는 {myCar.speed}km/hr")
```

## 5. 클래스 상속(inheritance)
    1) 기존 클래스의 속성과 메소드를 그대로 물려받는 새로운 클래스를 만드는 것
    2) 기존 클래스에 기능을 추가하거나 변경하여 새로운 클래스 정의 가능
    3) 코드 재사용 가능 ex) 자동차 -> 승용차, 트럭 클래스 상속
    4) 기존 클래스는 Parent, Super, Base 클래스로 분류
    5) 상속받는 클래스는 Child, Sub, Derived 클래스라고 부름

```python
# 예시) Car 클래스를 상속 받는 Sedan, Truck 클래스 생성

class Car:
    color = " "
    speed = 0

    def __init__(self, color, speed):
        self.color = color
        self.speed = speed

    def upSpeed(self, value):
        self.speed += value
    def downSpeed(self, value):
        self.speed -= value

# Car 클래스를 상속 받은 Sedan 클래스
# 좌석 수의 속성 추가
class Sedan(Car):
    seatNum = 0
    def getSeatNum(self):
        return self.seatNum

# Car 클래스를 상속 받은 Truck 클래스
# 용적 속성 추가
class Truck(Car):
    capacity = 0 
    def getCapacity(self):
        return self.capacity

myCar = Car('빨강', 10)
sedan = Sedan('노랑', 40)
truck = Truck('흰색', 100)
sedan.upSpeed(50)
truck.downSpeed(30)
sedan.seatNum = 5
truck.capacity = 500

print(f"Sedan은 {sedan.color}색이고, 속도는 {sedan.speed}km/hr이며, 좌석수는 {sedan.seatNum}개이다")
print(f"Truck은 {truck.color}색이고, 속도는 {truck.speed}km/hr이며, 용적은 {truck.capacity}이다")
```

## 6. 메소드 오버라이딩(overriding)
    1) 부모 클래스의 메소드를 하위 클래스에서 재정의하는 것
    2) 자식 클래스는 메소드를 상속받았기 때문에 다시 메소드를 생성할 필요 없음
    3) 자식 클래스는 메소드의 내용을 수정해야 한다면 메소드를 다시 만들 수 있음
    4) 자식 클래스의 객체로 호출시, 재정의된 메소드가 호출됨

```python
# Car 클래스를 상속받은 Sedan 클래스의 속도에 대한 메소드 재정의
# Sedan의 upSpeed 메소드의 속도제한 재정의

class Car:
    color = " "
    speed = 0

    def __init__(self, color, speed):
        self.color = color
        self.speed = speed

    def upSpeed(self, value):
        self.speed += value
    def downSpeed(self, value):
        self.speed -= value

class Sedan(Car):
    seatNum = 0
    def getSeatNum(self):
        return self.seatNum
    def upSpeed(self, value):
        self.speed += value
        if self.speed > 150:
            self.speed =150

sedan = Sedan('노랑', 50)
sedan.upSpeed(200)       #upSpeed를 200을 했지만 overriding에 의해 속도 150까지로 제한
sedan.seatNum = 5
print(f"Sedan은 {sedan.color}색이고, 속도는 {sedan.speed}km/hr이며, 좌석수는 {sedan.seatNum}개이다")
```

# 용어 정리: overwrite · override · overload 정리

| 용어 | 설명 |
| --- | --- |
| overwrite (덮어쓰기) | 기존 값·파일·정의를 새것으로 **교체**하는 것. 기존 것은 사라짐<br>예: 변수 재대입, `'w'` 모드로 파일 열기, 같은 이름의 함수 재정의 |
| override (재정의) | 상속받은 메서드를 자식 클래스에서 **같은 이름으로 다시 정의**하는 것<br>부모 메서드는 그대로 남아 있고 `super()`로 호출 가능<br>Java: 이름·매개변수·반환 타입이 같아야 함 / 파이썬: 이름만 같으면 됨 |
| overload (중복 정의) | **한 클래스 안에** 같은 이름, 다른 매개변수(개수·타입)의 메서드를 여러 개 두는 것<br>같은 기능을 매개변수에 따라 다르게 처리할 때 사용<br>Java·C++은 지원 / 파이썬은 미지원(나중 정의가 덮어씀) → 기본값 매개변수나 `*args`로 대신함 |

## 한 줄 요약
- **overwrite**: 덮어써서 원래 것이 **사라짐**
- **override**: 부모 것은 **남겨 두고** 자식에서 다시 정의
- **overload**: 한 클래스 안에서 같은 이름을 매개변수만 다르게 **여러 개** 정의 (파이썬은 미지원)

```python
# overwrite (덮어쓰기) 기존 것이 새것으로 바뀌고, 원래 값은 남지 않는다.

# 1) 변수 덮어쓰기
speed = 100
speed = 150
print(speed)

# 2) 파일 덮어쓰기: 'w'모드는 기존 내용을 지우고 새로 씀
with open("memo.txt", "w", encoding="utf-8") as f:
    f.write("김환자 - 고혈압\n")
with open("memo.txt", "w", encoding="utf-8") as f:
    f.write("이환자 - 당뇨\n")
with open("memo.txt", "r", encoding="utf-8") as f:
    print(f.read()) # 이환자 - 당뇨 (김환자 기록은 사라짐)

# 3) 함수 덮어쓰기: 같은 이름을 다시 정의하면 나중 것만 남음
def greet() :
    return "안녕하세요."
def greet():
    return "반갑습니다."
print(greet()) # 반갑습니다로 덮어씀
```

```python

# override (재정의): 자식 클래스가 부모의 메서드를 같은 이름으로 다시 정의한다. 부모 클래스의 메서드는 바뀌지 않고, 자식 객체에서만 새 동작이 실행
class Car:
    def __init__(self, color, speed):
        self.color = color
        self.speed = speed

    def upSpeed(self, value):
        self.speed += value

class Sedan(Car):
    maxSpeed = 150

    def upSpeed(self, value):
        super().upSpeed(value)
        self.speed = min(self.speed, self.maxSpeed)

car = Car('빨강', 100)
sedan = Sedan('노랑', 100)

car.upSpeed(100)
sedan.upSpeed(100)

print(car.speed)  # 200 (부모 클래스에서 정의된 메서드 그대로 출력)
print(sedan.speed) # 150  (자식 클래스에서 재정의된 메서드 출력)
```

```python
# overload (중복 정의): 파이썬은 오버로딩을 지원하지 않아서, 같은 이름으로 두 번 정의하면 overwrite가 됩니다. 그래서 기본값 매개변수나 `*args`로 같은 효과를 냅니다.

# 파이썬에서 오버로딩을 시도하면 → 나중 정의가 앞의 것을 덮어씀
class CalcTry:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):                 # 위의 add(a, b)를 덮어씀
        return a + b + c

try:
    CalcTry().add(1, 2)
except TypeError as e:
    print("TypeError:", e)


# 파이썬다운 대안 1: 기본값 매개변수
class Calc:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calc()
print(calc.add(1, 2))       # 3
print(calc.add(1, 2, 3))    # 6


# 파이썬다운 대안 2: *args로 개수 제한 없이 받기
class FlexCalc:
    def add(self, *numbers):
        return sum(numbers)

print(FlexCalc().add(1, 2))          # 3
print(FlexCalc().add(1, 2, 3, 4))    # 10
```
