# 모듈

## 1. 모듈(module)
    1) 필요할 때 꺼내쓸 수 있게 별도 파일에 함수들을 모아 놓은 것, 함수의 집합
    2) 미리 만들어둔 함수를 다른 프로그램에서도 쓰고 싶을 때
        - 자주 쓰는 함수들을 모아서 별도 .py로 관리
        - 파일 이름은 어떤 것으로 지어도 상관없으나 기능을 알 수 있게 명명

## 2. 모듈 불러오기
    1) import 모듈명
        - 해당 모듈안에 있는 모든 함수를 불러올 수 있음
        - 모듈명.함수명()으로 메소드 함수 호출 기능
    2) from 모듈명 import 함수명
        - 해당 모듈에서 특정 함수만 import
        - 모듈명은 생략하고 함수 이름만으로 사용하고 싶을 때
    3) from 모듈명 import *
        - 해당 모듈에 저장된 모든 함수 import
    4) import 모듈명 as 별명
        - 모듈 import시 별명(alias) 지정 가능
    5) 대표적인 모듈
        - request: HTTP 요청/응답 모듈
        - numpy: 수치해석 모듈
        - pandas:  데이터 분석 모듈
        - math: 수학적 수식 모듈
        - 다양한 기능들이 미리 함수로 구현되어 모듈 형태로 제공됨
