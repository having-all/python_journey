## 🐍 Quiz 1 — 함수 실행 결과 예측하기
다음 코드를 실행했을 때 출력되는 결과를 작성하고, 그 이유를 간단히 설명하세요.

1. 최종 출력 결과를 정확하게 작성하세요.
2. append()의 역할을 설명하세요.
3. return과 print()의 차이를 설명하세요.

```python
def calculate(numbers):
    result = []

    for num in numbers:
        if num % 2 == 0:
            result.append(num * 2)
        else:
            result.append(num + 1)

    return result

numbers = [3, 6, 9, 12]
print(calculate(numbers))
```

# 결과 예측하기
1. 최종 출력 결과를 정확하게 작성
- numbers 리스트의 값의 숫자 하나씩 꺼내어 짝수에는 곱하기 2, 홀수에는 더하기 1을 한 값을 result 리스트에 담아준다.
- 예측 결과: [4, 12, 10, 24]

2. append()의 역할을 설명하세요.
- 리스트의 마지막에 append 안의 값을 추가한다

3. return과 print()의 차이를 설명하세요
- return: 함수 안에서 값을 반환
- print: 값을 출력

**AI의 return과 print 차이 설명**
- return: 함수 안에서 결과 값을 반환하고 그 값을 변수에 저장해 다시 사용 가능
- print: 값을 화면에 출력, 함수 안에서 print만 하고 return이 없으면 그 함수는 None을 돌려준다

## 🐍 Quiz 2 — 클래스 + 딕셔너리 종합 문제
다음은 상품의 이름과 가격을 관리하는 ProductManager 클래스 입니다.   
상품 목록에서 가격이 10,000원 이상인 상품만 찾아 할인된 가격을 딕셔너리로 반환하는 메서드를 완성하세요.  

**문제 조건**
1. get_discounted_products() 메서드를 완성하세요.
2. 가격이 10,000원 이상인 상품만 선택하세요.
3. 선택된 상품의 가격에 10% 할인을 적용하세요.
4. 결과를 {상품명: 할인가격} 형태의 딕셔너리로 반환하세요.
5. for, if, return을 사용하세요.
6. 할인 가격은 int()를 이용해 정수로 변환하세요.

```python
class ProductManager:
    def __init__(self, products):
        self.products = products

    def get_discounted_products(self):
        choices = {}                                           # 가격이 10000 이상인 상품만 담을 딕셔너리
        for product in self.products:
            if product["price"] >= 10000:
                discounted = int(product["price"] * 0.1)
                choices[product["name"]] = product["price"]    # choices 딕셔너리의 키(상품명)에 value(가격)을 지정해줌
        return choices

products = [
    {"name": "Keyboard", "price": 30000},
    {"name": "Mouse", "price": 8000},
    {"name": "Monitor", "price": 150000},
    {"name": "USB", "price": 5000},
    {"name": "Speaker", "price": 20000}
]

manager = ProductManager(products)

print(manager.get_discounted_products())

```
