## Q1. 쇼핑몰 주문 데이터 분석 ⭐⭐⭐
**업무 상황:** 결제 완료 주문만 합산하여 총매출을 계산합니다.

**조건:** `status == "completed"`만 집계, 매출은 `price * quantity`, `sum()` 사용 금지.

**기대 결과:** `106000`

```python
orders = [
    {"id": 1, "price": 15000, "quantity": 2, "status": "completed"},
    {"id": 2, "price": 30000, "quantity": 1, "status": "cancelled"},
    {"id": 3, "price": 12000, "quantity": 3, "status": "completed"},
    {"id": 4, "price": 25000, "quantity": 1, "status": "refunded"},
    {"id": 5, "price": 8000, "quantity": 5, "status": "completed"}
]

def calculate_revenue(orders):
    total = 0
    for order in orders:
        if order["status"] == "completed":
            total += order["price"] * order["quantity"]
    return total
print(calculate_revenue(orders))
```

## Q2. 회원 데이터 중복 제거 ⭐⭐⭐
**업무 상황:** 이메일 기준으로 이벤트 신청자 중복을 제거합니다.

**조건:** 첫 등장한 회원만 유지, 기존 순서 유지, 리스트 반환.

**기대 결과:** Kim, Lee, Choi 세 명의 정보만 남습니다.

```python
users = [
    {"name": "Kim", "email": "kim@test.com"},
    {"name": "Lee", "email": "lee@test.com"},
    {"name": "Park", "email": "kim@test.com"},
    {"name": "Choi", "email": "choi@test.com"},
    {"name": "Jung", "email": "lee@test.com"}
]

def remove_duplicates(users):
    removed = []
    seen = set()
    for user in users:
        email = user["email"]
        if email not in seen:
            seen.add(email)         # set은 append 사용할 수 없다 => add사용
            removed.append(user)
    return removed

print(remove_duplicates(users))
```

## Q3. 서버 로그에서 오류 발생 횟수 집계 ⭐⭐⭐⭐
**업무 상황:** 서버 로그에서 오류 유형별 발생 횟수를 집계합니다.

**조건:** `ERROR:`로 시작하는 로그만 처리, `startswith()`와 `split()` 활용, `{메시지: 횟수}` 반환.

**기대 결과:** `{"Database connection failed": 2, "Timeout": 3}`

```python
logs = [
    "INFO: Server started",
    "ERROR: Database connection failed",
    "INFO: User logged in",
    "ERROR: Timeout",
    "ERROR: Database connection failed",
    "WARNING: High memory usage",
    "ERROR: Timeout",
    "ERROR: Timeout"
]

def count_errors(logs):
    errors = {}
    for log in logs:
        if log.startswith("ERROR:"):
            message = log.split(":", 1)[1].strip()
            if message in errors:
                errors[message] += 1
            else:
                errors[message] = 1
    return errors

print(count_errors(logs))
```

### Q1 정답 — 매출 집계
결제 완료 주문만 선택해 `가격 × 수량`을 누적합니다.

```python
def calculate_revenue(orders):
    total = 0
    for order in orders:
        if order["status"] == "completed":
            total += order["price"] * order["quantity"]
    return total

assert calculate_revenue(orders) == 106000
print(calculate_revenue(orders))
```

### Q2 정답 — 중복 제거
`set`에 처리한 이메일을 저장해 중복을 빠르게 확인합니다.

```python
def remove_duplicates(users):
    result = []
    seen = set()
    for user in users:
        email = user["email"]
        if email not in seen: 
            seen.add(email)
            result.append(user)
    return result

assert [u["name"] for u in remove_duplicates(users)] == ["Kim", "Lee", "Choi"]
print(remove_duplicates(users))
```

### Q3 정답 — 로그 분석
오류 로그를 필터링한 뒤 콜론 뒤 메시지를 추출하여 횟수를 누적합니다.

```python
def count_errors(logs):
    result = {}
    for log in logs:
        if log.startswith("ERROR:"):
            message = log.split(":", 1)[1].strip()
            if message in result:
                result[message] += 1
            else:
                result[message] = 1
    return result

assert count_errors(logs) == {"Database connection failed": 2, "Timeout": 3}
print(count_errors(logs))
```

### Q4 정답 — 재고 관리
재고를 변경하기 **전에** 출고 수량을 검증합니다.

```python
class Inventory:
    def __init__(self, product, stock):
        self.product = product
        self.stock = stock

    def release(self, quantity):
        if quantity <= 0:
            return False
        if quantity > self.stock:
            return False
        self.stock -= quantity
        return True

item = Inventory("Keyboard", 10)
assert item.release(7) is True
assert item.release(5) is False
assert item.release(0) is False
assert item.stock == 3
print(item.stock)
```

### Q5 정답 — 고객 등급 분류
금액이 큰 기준부터 검사하고 `copy()`로 원본 딕셔너리 변경을 방지합니다.

```python
def classify_customers(customers):
    result = []
    for customer in customers:
        total = customer["total"]
        if total >= 500000:
            grade = "VIP"
        elif total >= 200000:
            grade = "GOLD"
        elif total >= 100000:
            grade = "SILVER"
        else:
            grade = "BASIC"
        new_customer = customer.copy()
        new_customer["grade"] = grade
        result.append(new_customer)
    return result

result = classify_customers(customers)
assert [c["grade"] for c in result] == ["VIP", "GOLD", "SILVER", "BASIC", "VIP"]
assert all("grade" not in c for c in customers)
print(result)
```
