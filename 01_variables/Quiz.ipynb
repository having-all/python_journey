# BMI 계산하기

> height_cm와 weight 변수로 BMI를 계산해 bmi 변수에 담고, 소수점 2자리로 BMI: 23.05 형식으로 출력하세요.

```python
height_cm = 172.4
weight = 68.5

# 코드 작성
```

```python
# 모범 답안
height_cm = 172.4
weight = 68.5

bmi = weight / (height_cm / 100) ** 2
print(f"BMI: {bmi:.2f}")
```

# 약 포장 계산하기

> 알약 total_pills개를 per_pack개씩 포장합니다. 완성 포장 수와 남는 알약 수를 full_packs, remaining 변수에 구하고 완성 포장: 6봉지, 남는 알약: 3개 형식으로 출력하세요.

```python
total_pills = 45
per_pack = 7

# 여기에 작성
```

```python
# 모범 답안

total_pills = 45
per_pack = 7

full_packs = total_pills // per_pack
remaining = total_pills % per_pack
print(f"완성 포장: {full_packs}봉지, 남는 알약: {remaining}개")
```

# 고위험군 판정하기

```python
age = 70
has_fever = False
spo2 = 93

# 여기에 작성
```

```python
# 모범 답안
age = 70
has_fever = False
spo2 = 93

high_risk = age >= 65 and (has_fever or spo2 < 95)
print(f"고위험군: {high_risk}")
```
