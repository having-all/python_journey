# 연습문제
1. 환자 기본정보 저장 및 계산하기
> 아래 정보를 변수에 저장하고 BMI, 현재 나이, 성인 여부(18세 이상)를 계산해 출력하세요. 
> 현재 연도는 from datetime import datetime 후 datetime.now().year로 구합니다.
• 이름 "박건강", 출생연도 1990, 키 170cm, 몸무게 65kg, 성별 "남성"

출력예시)
```
=== 환자 정보 ===
이름: 박건강 (남성, 36세)
키: 170cm, 몸무게: 65kg
BMI: 22.5
출생연도: 1990년
성인 여부: True
```

````python
from datetime import datetime

name = '박건강'
gender = '남성'
height = 170
weight = 65
birth_year = 1990

# 주어진 정보로 age, bmi 계산
now = datetime.now()
age = now.year - birth_year
bmi = weight / (height/100)**2
is_adult = age >= 18


# 환자정보 출력
print("=== 환자정보 ===")
print(f"이름: {name} ({gender}, {age}세)")
print(f"키: {height}cm, 몸무게: {weight}kg")
print(f"BMI: {bmi:.1f}")
print(f"출생연도: {age}")
print(f"성인 여부: {is_adult}")
````
