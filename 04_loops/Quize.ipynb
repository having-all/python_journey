# 문자열 거꾸로 만들기

> reverse_text(text)가 문자열을 거꾸로 뒤집어 return 하도록 for문으로 작성하세요. (슬라이싱 [::-1]을 써도 통과합니다)

````python
# text slicing을 이용

text = "Hello, World!"

def reverse_text(text):
    reverse = ""
    for s in text[::-1]:   # text[::-1] -> 처음부터 끝까지 뒤로 한칸씩
        reverse += s
    return reverse

print(reverse_text(text))
````

````python
# range를 이용

text = "Hello, World!"

def reverse_text(text):
    reverse = ""
    for s in range(len(text) - 1, -1, -1):   # range(len(text) -1): 문자열의 인덱스 0부터 시작, 마지막 인덱스는 갯수 -1 임
        reverse += text[s]
    return reverse

print(reverse_text(text))
````
