```python
# for문 기본

medications = ["아스피린", "메트포르민", "암로디핀"]

for med in medications:      # medications 리스트의 값 1개씩을 반복한다, 반드시 : 입력
    print(f"{med} 복용 확인")
print("모든 약 복용 확인 완료")  # for문이 다 끝나고 반환
```

```python
# range: 101호부터 105호까지
for room in range(101, 106):
    print(f"{room}호 병실 점검 완료")

# enumerate: 1번부터 번호 붙이기
waiting = ["김환자", "이환자", "박환자"]

for num, patient in enumerate(waiting, 1):  # 숫자 1은 번호를 매기기 시작하는 수
    print(f"대기 {num}번: {patient}")

# zip: 3가지 리스트를 동시에
names = ["김환자", "이환자"]
drugs = ["해열제", "항생제"]
dosages = ["1일 3회", "1일 2회"]

for name, drug, dosage in zip(names, drugs, dosages):
    print(f"{name}: {drug} ({dosage})")
```

```python
# 딕셔너리 순회
vitals = {"bp": 158, "temp": 38.3, "hr": 78}
normals = {"bp": (90, 139), "temp": (36.1, 37.2), "hr": (60, 100)}

for key, value in vitals.items():
    low, high = normals[key]
    status = "정상" if low <= value <= high else "이상"
    print(f"{key}: {value} -> {status}")
```

```python
# 반복문의 3가지 패턴
temperatures = [36.5, 38.2, 37.1, 39.0, 36.8]

# 1. 조건에 맞는 것만 처리
for temp in temperatures:
    if temp >= 37.5:
        print(f"{temp}도: 발열")

# 2. 누적 카운트 합계
fever_count = 0
total = 0
for temp in temperatures:
    total += temp
    if temp >= 37.5:
        fever_count += 1
print(f"발열 {fever_count}명, 평균: {total/ len(temperatures):.1f}")

# 3. 새 리스트에 모으기
fever_list = []
for temp in temperatures:
    if temp >= 37.5:
        fever_list.append(temp)
print(f"발열: {fever_list}")
```

```python
# while 문

bp = 180
target = 150
max_dose = 5
dose_count =0

while bp > target and dose_count < max_dose:
    bp -= 15
    dose_count += 1
    print(f"{dose_count}회 투약 후: {bp}mmHg")

if bp <= target:
    print(f"목표 달성 ({dose_count}회 투약)")
else:
    print("최대 횟수 초과 => 의사 호출")
```
