## 연습 문제
> 주어진 patients_db를 가지고 다음을 순서대로 처리하세요
```
1. 첫 번째 환자 정보를 get()으로 조회해 출력
2. 두 번째 환자의 질병을 update()로 "급성맹장염"으로 수정하고 전후 비교
3. 새 환자(P004, 홍환자, 40세, 피부과, 아토피)를 append()로 등록하고 patients_db[-1]로 다시 조회
4. 마지막 환자를 pop()으로 퇴원 처리
5. 반복문 없이 max, min, sum으로 나이 통계(최고령 65세, 최연소 32세, 평균 47.3세) 출력
```

````python
patients_db = [
    {"patient_id": "P001", "name": "김환자", "age": 65, "department": "내과", "disease": "고혈압"},
    {"patient_id": "P002", "name": "이환자", "age": 45, "department": "외과", "disease": "맹장염"},
    {"patient_id": "P003", "name": "박환자", "age": 32, "department": "소아과", "disease": "감기"},
]
````

````python
# 1.첫 번째환자 정보를 get()으로 조회해 출력
first_patient = patients_db[0]
print(first_patient.get("patient_id"))
print(first_patient.get("name"))
print(first_patient.get("age"))
print(first_patient.get("department"))
print(first_patient.get("disease"))
````

````python
# 2. 두 번째 환자의 질병을 update()로 "급성맹장염"으로 수정하고 전후 비교
second_patient = patients_db[1]

print(f"=== 두 번째 환자의 질병 update 수정 전 ===")
print(second_patient)

# update는 바꿀 내용을 딕셔너리 형태로 괄호안에 넣어야 한다.
second_patient.update({"disease":"급성맹장염"})
# 다른 방법
# second_patient["disease"] = "급성맹장염"
print(f"\n=== 두 번째 환자의 질병 update 수정 후 === ")
print(second_patient)
````

````python
# 3. 새 환자(P004, 홍환자, 40세, 피부과, 아토피)를 append()로 등록하고 patients_db[-1]로 다시 조회
# 새 환자의 정보를 new_patient 변수에 할당해준다
new_patient = {
    "patient_id": "P004",
    "name": "홍환자",
    "age": 40,
    "department": "내과",
    "disease": "아토피"
}
# append()로 new_patient 정보를 추가한다.
patients_db.append(new_patient)
print(f"새 환자: {patients_db[-1]}")
````

````python
# 4. 마지막 환자를 pop()으로 퇴원 처리
discharge_patient = patients_db.pop()
print("=== 퇴원 환자 정보 ===")
print(discharge_patient["patient_id"])
print(discharge_patient["name"])
print(discharge_patient["age"])
print(discharge_patient["department"])
print(discharge_patient["disease"])
````

````python
# 5. 반복문 없이 max, min, sum으로 나이 통계(최고령 65세, 최연소 32세, 평균 47.3세) 출력
ages = patients_db[0]["age"], patients_db[1]["age"], patients_db[2]["age"]
print("=== 나이 통계 ===")
print(f"최고령: {max(ages)}세")
print(f"최연소: {min(ages)}세")
print(f"평균 나이: {sum(ages)/len(ages):.1f}세")
````
