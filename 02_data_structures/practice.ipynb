````python
# List Method
a = [1, 2, 3]
a.append([4, 5])  # append는 뒤로 [4, 5]를 한 덩어리로 추가
print(a)
b = [1, 2, 3]
b.extend([4, 5])  # extend는 b리스트에 [4, 5] 리스트로 추가
print(b)

bp = [120, 125, 130]
print("/".join(map(str, bp)))
````

````python
# 튜플 ( ): 순서있음, 변경 불가 -> 바뀌면 안되는 정상 범위
BP_NORMAL = (90, 139)    # 정상 혈압 범위
low, high = BP_NORMAL    # 언패킹: 튜플 값을 여러 변수에 한 번에
bp = 145
print(low <= bp <= high)

# 언패킹: 튜플 값을 여러 변수에 한 번에
name, systolic, diastolic, temp = ("김환자", 130, 80, 36.7)
print(name)

BP_NORMAL[0] = 80        # 값을 변경할 수 없다
````

````python
# 딕셔너리 {키:값}: 이픔표로 찾기 -> 환자 챠트
patient = {"name":"김환자", "age":65,"diagnosis": "당뇨"}
print(patient)

patient["doctor"] = "이의사"  # 없는 키는 추가
patient["age"] = 66 # 있는 키는 수정
patient.update({"diagnosis":"고혈압, 당뇨"}) # 여러개 한번에 수정
removed = patient.pop("doctor")

print(patient.keys())
print(patient.values())
print(patient.items())

# 없는 키를 꺼내면 keyError 발생 -> get으로 안전하게 꺼내기
print(patient.get("보험번호")) # 값이 없어서 None으로 출력
````

````python
# 딕셔너리 중첩구조
chart = {
    "name" : "박환자",
    "vitals" : {"bp" : 150, "temp" : 37.2},
    "medications" :["아스피린", "암로디핀"]
}
print(chart["vitals"])
print(chart["vitals"]["bp"])      # chart 딕셔너리의 vital 키의 값 중 bp키의 값을 출력
print(chart["medications"][0])    # chart 딕셔너리의 medications 키의 값 중 1번째 값 출력

````

````python
# Set
symptoms = set(["발열", "기침", "발열", "두통"]) # set으로 "발열" 중복 값 하나만 출력
print(symptoms) 

allergies = {"페니실린", "아스피린", "설파제"}
prescription = {"아목사실린", "아스피린", "메트포르민"}
print(allergies & prescription)
print(prescription - allergies)

allergies.add("이부프로펜")
allergies.discard("없는 약")
print(allergies)
````
