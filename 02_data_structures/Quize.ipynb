#  체온 통계 구하기

> temps 리스트에서 평균(소수점 1자리 반올림), 최고, 최저 체온을 avg_temp, max_temp, min_temp 변수에 구하세요.

```python
temps = [36.5, 37.8, 38.2, 36.9, 37.1]

# 여기에 작성
```

```python
# 모범 답안
temps = [36.5, 37.8, 38.2, 36.9, 37.1]

avg_temp = round(sum(temps) / len(temps), 1)
max_temp = max(temps)
min_temp = min(temps)
print(avg_temp, max_temp, min_temp)
```

# 처방 안전성 검사

> 알레르기 약물과 새 처방이 겹치는 약을 conflict(집합)에, 알레르기를 뺀 안전한 처방을 가나다순 리스트 safe_rx에 구하세요.

```python
allergies = {"페니실린", "아스피린", "설파제"}
new_rx = {"아목시실린", "아스피린", "메트포르민", "설파제"}

# 여기에 작성
```

```python
# 모범 답안
allergies = {"페니실린", "아스피린", "설파제"}
new_rx = {"아목시실린", "아스피린", "메트포르민", "설파제"}

conflict = allergies & new_rx
safe_rx = sorted(new_rx - allergies)
print(conflict, safe_rx)
```

# 환자 차트 수정하기

> patient 딕셔너리에 담당의 "doctor": "박의사"를 추가하고, 나이를 61로 바꾸세요. 그리고 없는 키 "insurance"를 get()으로 꺼내 없으면 "미등록"이 되도록 insurance 변수에 담으세요.

```python
patient = {"name": "최환자", "age": 60, "diagnosis": "당뇨"}

# 여기에 작성
```

```python
# 모범 답안
patient = {"name": "최환자", "age": 60, "diagnosis": "당뇨"}

patient["doctor"] = "박의사"
patient["age"] = 61
insurance = patient.get("insurance", "미등록")
print(patient, insurance)
```
